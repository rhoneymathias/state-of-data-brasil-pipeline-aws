## Reconciliação de Schema entre os 3 anos da pesquisa

**Problema encontrado:** os nomes das colunas mudam de formato e de código
entre 2023, 2024 e 2025-2026, mesmo quando a pergunta é exatamente a mesma.

Exemplo real (pergunta "Gênero"):
- 2023: `('P1_b ', 'Genero')` — vem como texto de tupla Python
- 2024: `1.b_genero`
- 2025-2026: `1.b_genero`

**Por que isso importa:** se eu tentar juntar (`concat`) as 3 bases direto,
o pandas não reconhece essas colunas como "a mesma pergunta" — ele cria
colunas separadas e a base final fica cheia de buracos. Preciso primeiro
"traduzir" os nomes de coluna pra um padrão único antes de unir os anos.

**Exemplos: Procurar a coluna de gênero**

```python
# procura, em cada lista de colunas, qualquer nome que contenha a palavra "genero"
colunas_2023 = [c for c in dataset_2023.columns if "genero" in c.lower()]
colunas_2024 = [c for c in dataset_2024.columns if "genero" in c.lower()]
colunas_2025 = [c for c in dataset_2025.columns if "genero" in c.lower()]

print("2023:", colunas_2023)
print("2024:", colunas_2024)
print("2025:", colunas_2025)
```

**Separar código de texto, na coluna.**

Pegar as variáveis (nome_coluna_2023, nome_coluna_2024, nome_coluna_2025) e quebrar cada uma em duas partes: codigo e texto.

2024 e 2025 primeiro, porque é mais simples — o separador é o primeiro _:

```python
codigo_2024, sep, texto_2024 = nome_coluna_2024.partition("_")
print("codigo:", codigo_2024)
print("texto:", texto_2024)

codigo_2025, sep, texto_2025 = nome_coluna_2025.partition("_")
print("codigo:", codigo_2025)
print("texto:", texto_2025)
```

**2023 é mais chato porque o nome vem como texto de tupla — "('P1_b ', 'Genero')".**


Aqui não dá pra usar .partition() simples, precisamos de uma expressão regular pra "ler" essa estrutura:

```python
import re

match = re.match(r"^\('([^']*)',\s*'(.*)'\)$", nome_coluna_2023)
codigo_2023 = match.group(1).strip()
texto_2023 = match.group(2).strip()
print("codigo:", codigo_2023)
print("texto:", texto_2023)
```

**Agora generalizamos com um loop — a mesma lógica, aplicada a todas as colunas de uma vez:**

```python
# Generaliza o que fizemos manualmente pra "genero" -> agora para TODAS as colunas
# de TODOS os anos, de uma vez, guardando o resultado numa tabela.

import re
import unicodedata

def parse_coluna_2023(c):
    # mesma regex que usamos manualmente, agora dentro de uma função reutilizável
    match = re.match(r"^\('([^']*)',\s*'(.*)'\)$", c, re.DOTALL)
    if match:
        return match.group(1).strip(), match.group(2).strip()
    return c, c  # fallback, caso alguma coluna não bata no padrão esperado

def parse_coluna_outros(c):
    # mesmo .partition("_") que usamos manualmente pra 2024/2025
    codigo, sep, texto = c.partition("_")
    return (codigo, texto) if sep else (c, c)

def normaliza(texto):
    texto = texto.lower().strip()
    texto = unicodedata.normalize('NFKD', texto).encode('ascii', 'ignore').decode('ascii')
    # NOVO: troca underscore, barra, vírgula e outros separadores por espaço,
    # depois colapsa espaços múltiplos em um só - assim "cargo_atual" e
    # "cargo atual" viram exatamente a mesma string
    texto = re.sub(r"[^a-z0-9]+", " ", texto).strip()
    return texto

# Aplica a função em CADA coluna de CADA ano - isso substitui o trabalho manual
registros = []
for coluna in dataset_2023.columns:
    codigo, texto = parse_coluna_2023(coluna)
    registros.append({"ano": "2023", "coluna_original": coluna, "codigo": codigo, "pergunta_normalizada": normaliza(texto)})

for coluna in dataset_2024.columns:
    codigo, texto = parse_coluna_outros(coluna)
    registros.append({"ano": "2024", "coluna_original": coluna, "codigo": codigo, "pergunta_normalizada": normaliza(texto)})

for coluna in dataset_2025.columns:
    codigo, texto = parse_coluna_outros(coluna)
    registros.append({"ano": "2025-2026", "coluna_original": coluna, "codigo": codigo, "pergunta_normalizada": normaliza(texto)})

df_todas_colunas = pd.DataFrame(registros)
print(f"Total de colunas processadas: {len(df_todas_colunas)}")
df_todas_colunas.head(10)
```

**Quais perguntas existem nos 3 anos (essas viram sua base confiável), quais existem em só 1 ou 2 (essas precisam de decisão manual, ou ficam de fora por enquanto)?**

```python
# agrupa por pergunta normalizada e conta em quantos ANOS DIFERENTES ela aparece
contagem_anos = df_todas_colunas.groupby("pergunta_normalizada")["ano"].nunique()

print("Distribuição — em quantos anos cada pergunta aparece:")
print(contagem_anos.value_counts().sort_index())

# isolamos as que aparecem nos 3 anos - essa é nossa base confiável
perguntas_completas = contagem_anos[contagem_anos == 3]
print(f"\nPerguntas presentes nos 3 anos: {len(perguntas_completas)}")
```

**Até agora:**

- 215 perguntas que existem, com o mesmo texto, nos 3 anos — essa é a base do seu schema unificado

# **Usar as 215 perguntas pra efetivamente juntar as 3 bases numa única tabela (isso é o que vira sua camada Silver).**

**Filtrar o mapa de colunas só pras 215 perguntas que vamos usar**

Temos o df_todas_colunas (as 1.190 linhas com ano/coluna_original/código/pergunta_normalizada) e perguntas_completas (as 215 que aparecem nos 3 anos). Agora filtramos a primeira tabela pra manter só as linhas relacionadas a essas 215:

```python
# pega só o texto normalizado das 215 perguntas que decidimos usar
perguntas_usadas = perguntas_completas.index

# filtra df_todas_colunas para manter só as linhas dessas perguntas
mapa = df_todas_colunas[df_todas_colunas["pergunta_normalizada"].isin(perguntas_usadas)]
print(f"Linhas no mapa filtrado: {len(mapa)}")
print(f"Deveria ser aproximadamente: {215 * 3} (215 perguntas x 3 anos, se cada uma aparecer 1x por ano)")
```

```python
# conta quantas colunas originais diferentes caem em cada combinação (ano, pergunta)
contagem_por_ano_pergunta = mapa.groupby(["ano", "pergunta_normalizada"]).size()

# isola só os casos com MAIS DE 1 coluna original -> isso é ambiguidade
ambiguos = contagem_por_ano_pergunta[contagem_por_ano_pergunta > 1]
print(f"Combinações ano+pergunta com ambiguidade (mais de 1 coluna original): {len(ambiguos)}")
ambiguos.head(15)
```

771 > 645 confirma a previsão, e as 111 combinações ambíguas mostram exatamente o padrão que eu esperava: nomes de ferramentas/tecnologias (Apache Airflow, Databricks, Alteryx, AWS Glue...) que aparecem repetidos em blocos diferentes da pesquisa (ex: "ferramentas que uso hoje" vs "ferramentas que gostaria de aprender" — são perguntas diferentes, coincidência de texto).

**Separar as perguntas seguras das ambíguas**

Regra de decisão: se uma pergunta tem ambiguidade em qualquer um dos 3 anos, ela sai do lote automático — mesmo que nos outros 2 anos ela seja limpa. Não dá pra confiar parcialmente numa pergunta.

```python
# pega os nomes de pergunta que aparecem no índice de "ambiguos" (independente do ano)
perguntas_ambiguas = ambiguos.index.get_level_values("pergunta_normalizada").unique()
print(f"Perguntas com ambiguidade em pelo menos 1 ano: {len(perguntas_ambiguas)}")

# as perguntas seguras = as 215 que decidimos usar, MENOS as ambiguas
perguntas_seguras = [p for p in perguntas_usadas if p not in perguntas_ambiguas]
print(f"Perguntas seguras para o merge automático: {len(perguntas_seguras)}")
```

# **Montar a tabela final unificada (Silver)**

Agora vem a parte que efetivamente junta os 3 anos. O mecanismo, em 3 movimentos por ano:

Pra cada ano, olhar cada coluna original e descobrir se ela corresponde a uma das 172 perguntas seguras.
Se sim, guardar um "mapa de renomeação": {nome_original: nome_canônico}.
Selecionar só essas colunas do dataset daquele ano, renomear, e adicionar uma coluna ano_pesquisa pra saber de onde cada linha veio depois que tudo estiver empilhado.

```python
def monta_renome(dataset, ano, parser):
    """
    Para um dataset e seu parser (parse_coluna_2023 ou parse_coluna_outros),
    devolve um dicionario {nome_coluna_original: nome_canonico},
    só para as colunas que caem numa das perguntas_seguras.
    """
    renome = {}
    for coluna_original in dataset.columns:
        codigo, texto = parser(coluna_original)
        pergunta = normaliza(texto)
        if pergunta in perguntas_seguras:
            renome[coluna_original] = pergunta
    return renome

# monta o mapa de renomeação pra cada ano, usando o parser certo pra cada um
renome_2023 = monta_renome(dataset_2023, "2023", parse_coluna_2023)
renome_2024 = monta_renome(dataset_2024, "2024", parse_coluna_outros)
renome_2025 = monta_renome(dataset_2025, "2025-2026", parse_coluna_outros)

print(f"2023: {len(renome_2023)} colunas mapeadas")
print(f"2024: {len(renome_2024)} colunas mapeadas")
print(f"2025-2026: {len(renome_2025)} colunas mapeadas")
```

**Selecionar, renomear e empilhar (o merge de verdade)**

Agora usamos os 3 dicionários de renomeação pra criar 3 tabelas "traduzidas" pro mesmo padrão de nomes, e depois empilhar elas (uma embaixo da outra, não lado a lado) numa única tabela.

```python
def prepara_ano(dataset, renome, ano):
    """
    Seleciona só as colunas mapeadas, renomeia pro nome canônico,
    e marca de qual ano cada linha veio.
    """
    colunas_originais = list(renome.keys())
    df_selecionado = dataset[colunas_originais].rename(columns=renome)

    # se por acaso 2 colunas originais caírem no mesmo nome canônico
    # (não deveria acontecer aqui, já filtramos as ambíguas - mas checar é grátis)
    df_selecionado = df_selecionado.loc[:, ~df_selecionado.columns.duplicated()]

    df_selecionado["ano_pesquisa"] = ano
    return df_selecionado

tabela_2023 = prepara_ano(dataset_2023, renome_2023, "2023")
tabela_2024 = prepara_ano(dataset_2024, renome_2024, "2024")
tabela_2025 = prepara_ano(dataset_2025, renome_2025, "2025-2026")

print("Formato de cada tabela (linhas, colunas):")
print("2023:", tabela_2023.shape)
print("2024:", tabela_2024.shape)
print("2025-2026:", tabela_2025.shape)

# empilha as 3 tabelas - uma embaixo da outra (ignore_index reseta a numeração de linha)
df_silver = pd.concat([tabela_2023, tabela_2024, tabela_2025], ignore_index=True)

print("\nTabela final unificada (Silver):")
print(df_silver.shape)
```

**Recalcular 'genero' em cima da tabela nova e conferir se os percentuais batem com o que foi feito manualmente no começo** (75,1% / 76,1% / 77,5% masculino). Se bater, temos prova de que o merge não corrompeu nada.

```python
# recalcula genero por ano, agora usando a tabela unificada (Silver)
for ano in ["2023", "2024", "2025-2026"]:
    sub = df_silver[df_silver["ano_pesquisa"] == ano]
    distrib = sub["genero"].value_counts(normalize=True) * 100
    print(f"\n{ano} (n={len(sub)}):")
    print(distrib.round(1))
```

```python
from scipy.stats import chi2_contingency

tabela_contingencia = pd.crosstab(df_silver["ano_pesquisa"], df_silver["genero"])
print(tabela_contingencia)

chi2, p_valor, gl, esperado = chi2_contingency(tabela_contingencia)
print(f"\np-valor: {p_valor:.4f}")
# convenção: p-valor < 0.05 -> a diferença entre anos é estatisticamente significativa,
# não é só ruido de amostra
```

# **Remuneração × Senioridade**

```python
# 1. achar a coluna de faixa salarial e ver os valores existentes
colunas_salario = [c for c in df_silver.columns if "salari" in c]
print("Colunas encontradas:", colunas_salario)

coluna_salario = colunas_salario[0]
print("\nValores únicos de faixa salarial:")
print(df_silver[coluna_salario].value_counts(dropna=False))
```

```python
# 2. cria uma versão numerica aproximada (ponto medio de cada faixa) -
# fica mais facil comparar "salario medio" entre grupos do que ficar
# olhando 13 categorias de texto lado a lado
ponto_medio = {
    "Menos de R$ 1.000/mês": 500,
    "de R$ 1.001/mês a R$ 2.000/mês": 1500,
    "de R$ 2.001/mês a R$ 3.000/mês": 2500,
    "de R$ 3.001/mês a R$ 4.000/mês": 3500,
    "de R$ 4.001/mês a R$ 6.000/mês": 5000,
    "de R$ 6.001/mês a R$ 8.000/mês": 7000,
    "de R$ 8.001/mês a R$ 12.000/mês": 10000,
    "de R$ 12.001/mês a R$ 16.000/mês": 14000,
    "de R$ 16.001/mês a R$ 20.000/mês": 18000,
    "de R$ 20.001/mês a R$ 25.000/mês": 22500,
    "de R$ 25.001/mês a R$ 30.000/mês": 27500,
    "de R$ 30.001/mês a R$ 40.000/mês": 35000,
    "Acima de R$ 40.001/mês": 45000,
}
# as 2 linhas malformadas (N=1 cada) nao estao no dicionario -> viram NaN automaticamente, saem da conta
df_silver["salario_estimado"] = df_silver["faixa salarial"].map(ponto_medio)

# 3. salario medio estimado por nivel (usa mediana, mais robusta a outlier que media)
resumo_salario = df_silver.groupby("nivel")["salario_estimado"].agg(
    mediana="median", media="mean", n="count"
).reindex(["Júnior", "Pleno", "Sênior", "Especialista/Staff+"])
print("Salario estimado por nivel (R$):")
print(resumo_salario.round(0))

# 4. mesmo cruzamento, mas por ano tambem - pra ver se o gap entre niveis mudou
resumo_por_ano = df_silver.groupby(["ano_pesquisa", "nivel"])["salario_estimado"].median().unstack()
resumo_por_ano = resumo_por_ano.reindex(columns=["Júnior", "Pleno", "Sênior", "Especialista/Staff+"])
print("\nMediana salarial por nivel, por ano:")
print(resumo_por_ano.round(0))

# 5. grafico
import matplotlib.pyplot as plt
plt.figure(figsize=(7,4))
plt.bar(resumo_salario.index, resumo_salario["mediana"])
plt.title("Mediana salarial estimada por nível")
plt.ylabel("R$ (ponto médio da faixa)")
for i, v in enumerate(resumo_salario["mediana"]):
    plt.text(i, v+500, f"R$ {v:,.0f}", ha="center")
plt.savefig("/content/grafico_salario_por_nivel.png", dpi=150, bbox_inches="tight")
plt.show()
```

# **Tecnologia mais usada e diferença regional**

```python
# ===== TECNOLOGIA MAIS USADA =====
colunas_tech = ["python", "r", "sql", "scala", "julia", "rust", "c c c", "sql server"]

adocao_tech = {}
for c in colunas_tech:
    respondentes = df_silver[c].notna().sum()
    usa = df_silver[c].sum()
    adocao_tech[c] = round(usa / respondentes * 100, 1)

adocao_tech = pd.Series(adocao_tech).sort_values(ascending=False)
print("Adoção de linguagens (% de quem respondeu este bloco):")
print(adocao_tech)

# ===== DIFERENÇA REGIONAL =====
# regiao onde mora ja vem limpo, mas junto com estado - vamos usar a coluna de regiao mesmo
print("\nDistribuição por região:")
print(df_silver["regiao onde mora"].value_counts(dropna=False, normalize=True).mul(100).round(1))

# bonus rapido: regiao x nivel (mesmo padrao que ja fizemos, sem esforco extra)
print("\nSenioridade por região (%, dentro de cada região):")
print(pd.crosstab(df_silver["regiao onde mora"], df_silver["nivel"], normalize="index").mul(100).round(1))
```

1. **Tecnologia** — o insight real não é "SQL/Python dominam" (óbvio, esperado), é o tamanho do degrau depois deles. SQL e Python ficam acima de 80%, viraram tabela-base da profissão. Mas olha a queda: de Python (81,3%) pra SQL Server (31,3%) já é um degrau de 50 pontos, e daí pra R (11,5%) é outro tombo. Isso é sinal de mercado consolidado em torno de 2 ferramentas-padrão, com tudo o resto (R, Scala, Rust, Julia) sendo nicho/especialização, não tendência ampla. Pra capacitação, a leitura de negócio é: "domínio de Python+SQL é table stakes, não diferencial competitivo" — o investimento em treinamento que gera vantagem está nas ferramentas de nicho, não nas duas mais populares.

2. **Senioridade por região** — Sudeste tem a maior proporção de Sênior (39,3%) e a menor de Júnior (22,7%). Norte e Nordeste são o oposto: mais Júnior proporcionalmente (31,4% e 29,5%), menos Sênior. Isso sugere um padrão de "fuga de talento sênior" pro eixo Sudeste/Sul — regiões fora do eixo formam profissional júnior mas não retêm eles até virarem sênior ali.

   Isso é uma recomendação estratégica real pro cliente (a instituição financeira): se quiser contratar sênior fora do Sudeste, vai competir num pool bem mais raso; se quiser formar talento, Norte/Nordeste tem mais potencial de crescimento júnior→pleno disponível.
